```python
x = torch.tensor(2.0)
w = torch.tensor(3.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)
y = w * x + b
loss = (y - 10) ** 2
y, loss

```

```python
loss.backward()
print(w.grad.item())
print(b.grad.item())

```

```python
x = torch.tensor(3.0)
w = torch.tensor(4.0, requires_grad=True)
b = torch.tensor(6.0, requires_grad=True)
y = w * x + b
loss = (y - 10) ** 2
print(y)
print(loss)
loss.backward()
print(w.grad.item())
print(b.grad.item())

```

```python
x = torch.tensor(3.0)
w = torch.tensor(4.0, requires_grad=True)
b = torch.tensor(6.0, requires_grad=True)
y = w * x + b
loss = (y - 10) ** 2
dw, db = torch.autograd.grad(loss, (w, b))
dw, db

```
