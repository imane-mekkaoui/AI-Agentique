```python
import torch

```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
device

```

```python
x = torch.linspace(-2, 2, 50)
x

```

```python
x = x.unsqueeze(1)
x

```

```python
y = 3 * x + 2

```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 5))
plt.scatter(x.detach().numpy(), y.detach().numpy())
plt.show()

```

```python
noise = 0.6 * torch.randn_like(x)
y = 3 * x + 2 + noise
plt.figure(figsize=(15, 5))
plt.scatter(x.detach().numpy(), y.detach().numpy())
plt.show()

```

```python
from torch import nn

model = nn.Linear(1, 1)
model

```

```python
loss_function = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

```

```python
epochs = 500
history = []
model.train()
for epoch in range(epochs):
    optimizer.zero_grad()
    y_pred = model(x)
    loss = loss_function(y_pred, y)
    loss.backward()
    optimizer.step()
    history.append(loss.item())
    if epoch % 100 == 0:
        print(f"époque {epoch}/{epochs}  loss={loss.item()}")

```

```python
model.eval()
with torch.inference_mode():
    pred = model(x)

plt.figure(figsize=(15, 5))
plt.scatter(x.detach().numpy(), y.detach().numpy())
plt.plot(x.detach().numpy(), pred.numpy())
plt.show()

```

```python
from torchmetrics.regression import R2Score

r2_score = R2Score()
r2_score.update(pred, y)
print(r2_score.compute())

```

```python
plt.plot(history)
plt.show()

```

```python
class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(4, 16),
            nn.ReLU(),
            nn.Linear(16, 8),
            nn.ReLU(),
            nn.Linear(8, 3),
        )

    def forward(self, x):
        return self.net(x)


model = MLP()
model

```

```python
torch.manual_seed(0)
n = 150
c0 = torch.randn(n, 2) * 0.55 + torch.tensor([-1.4, -0.8])
c1 = torch.randn(n, 2) * 0.55 + torch.tensor([1.3, 1.0])
X = torch.cat([c0, c1])
y_cls = torch.cat([torch.zeros(n), torch.ones(n)]).long()

plt.figure(figsize=(5, 5))
plt.scatter(c0[:, 0], c0[:, 1])
plt.scatter(c1[:, 0], c1[:, 1])
plt.show()

```

```python
clf = nn.Sequential(
    nn.Linear(2, 16),
    nn.ReLU(),
    nn.Linear(16, 16),
    nn.ReLU(),
    nn.Linear(16, 2),
)
loss_fn = nn.CrossEntropyLoss()
opt = torch.optim.Adam(clf.parameters(), lr=0.01)

history_clf = []
clf.train()
for epoch in range(200):
    opt.zero_grad()
    logits = clf(X)
    loss = loss_fn(logits, y_cls)
    loss.backward()
    opt.step()
    history_clf.append(loss.item())
    if epoch % 50 == 0:
        print(epoch, loss.item())
print(history_clf[-1])

```

```python
clf.eval()
with torch.inference_mode():
    pred_cls = clf(X).argmax(dim=1)

acc = (pred_cls == y_cls).float().mean()
print(acc)

from torchmetrics.classification import BinaryConfusionMatrix

bcm = BinaryConfusionMatrix()
cm = bcm(pred_cls, y_cls)
print(cm)
plt.imshow(cm)
plt.show()

```
