## D — Synthèse SAV (aperçu)

Orchestrateur → dépanneur **ou** vérificateur → retour orchestrateur → END.


```python
from ap5.orchestrateur import construire_graphe_sav

sav = construire_graphe_sav()
ex1 = sav.invoke({
    "question": "Mon imprimante PX-200 affiche un voyant clignotant. Que signifie le pictogramme sur la photo jointe ?",
    "photo": "img_px200_01",
})
print("Ex1", ex1.get("categorie"), ex1.get("agent"))
print(ex1.get("reponse"), "\n")

ex2 = sav.invoke({
    "question": "Est-ce que mon imprimante est encore sous garantie ? Son numéro de série est PX200-0042.",
    "photo": "",
})
print("Ex2", ex2.get("categorie"), ex2.get("agent"))
print(ex2.get("reponse"))
```
